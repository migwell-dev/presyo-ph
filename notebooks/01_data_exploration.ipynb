# Inspection

```python
df = pd.read_csv("../data/food_prices.csv")

print("Shape", df.shape, end="\n")
print("Columns:")
print(df.columns.tolist())

display(df.head)
```

```python
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.describe(include="all").T
```

```python
print("Unique markets:", df["mkt_name"].nunique())

print("\nMarkets:")
print(df["mkt_name"].unique())

print("\nUnique geographic IDs:", df["geo_id"].nunique())
```

```python
print("Earliest date:", df["price_date"].min())
print("Latest date:", df["price_date"].max())

print("\nRecords per year:")
print(df.groupby("year").size())
```

```python
national = df[df["geo_id"] == "gid_phl_national_average"]
local = df[df["geo_id"] != "gid_phl_national_average"]

print("National average records:", len(national))
print("Local market records:", len(local))
print("Unique local markets:", local["mkt_name"].nunique())
```

```python
commodities = [
    "beans", "cabbage", "carrots", "eggs",
    "garlic", "onions", "potatoes", "rice", "tomatoes"
]

missing = df[commodities].isnull().mean().sort_values(ascending=False)

print(missing)
```

```python
commodity_summary = pd.DataFrame({
    "available": df[commodities].count(),
    "missing": df[commodities].isnull().sum(),
    "missing_pct": df[commodities].isnull().mean() * 100
})

display(commodity_summary.sort_values("available", ascending=False))
```

```python
print("Exact duplicate rows:", df.duplicated().sum())

print(
    "Duplicate geographic ID + date:",
    df.duplicated(subset=["geo_id", "price_date"]).sum()
)
```

```python
df["price_date"] = pd.to_datetime(df["price_date"])

date = "2025-02-01"

df.loc[
    df["price_date"] == date,
    ["adm1_name", "adm2_name", "mkt_name", "price_date", "rice", "currency"]
].sort_values("rice", ascending=False)
```

```python
monthly_rice = (
    df.groupby("price_date")["rice"]
    .mean()
    .reset_index()
)

monthly_rice
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_rice["price_date"],
    monthly_rice["rice"]
)

plt.title("Monthly Average Rice Price Estimates")
plt.xlabel("Date")
plt.ylabel("Price (PHP per kg)")
plt.grid(True)
plt.show()
```
